# Phase 2 - Source Data Understanding

## 2.1 Identify Source files

The purpose of this phase is to understand the structure, content, and quality of the source data before performing any cleaning or transformation.

The first step is to identify the raw source files provided for the project.

```python
from pathlib import Path

# Define the location of the raw source data
raw_data_path = Path("../data/raw")

# List all files in the raw data directory
for file in raw_data_path.iterdir():
    print(file.name)
```

## 2.2 Source File Inventory

A source file inventory is created to understand the number, type, and size of
datasets available before loading them into memory.

```python
# create an inventory of the raw source files

source_files = list(raw_data_path.glob("*.csv"))

print(f"Number of CSV files: {len(source_files)}")
print()

for file in source_files:
    size_mb = file.stat().st_size / (1024 * 1024)
    print(f"{file.name:<45} {size_mb:.2f} MB")
```

## 2.3 Load Source Data

The CSV source files are loaded into Pandas DataFrames to enable structural
inspection and data profiling.

At this stage, the source data is loaded without applying transformations so that
the original structure and quality of the datasets can be assessed.

```python
import pandas as pd

# Load all CSV files into a dictionary of DataFrames
datasets = {}

for file in source_files:
    dataset_name = file.stem
    datasets[dataset_name] = pd.read_csv(file)

print(f"Successfully loaded {len(datasets)} datasets.")
datasets.keys()
```

```python
datasets["customers_dataset"].head()
```

## 2.4 Dataset Dimensions

The number of rows and columns in each dataset is reviewed to understand the
volume and structure of the source data.

```python
for name, df in datasets.items():
    print(f"{name:<40} Rows: {df.shape[0]:>8,} | Columns: {df.shape[1]:>2}")
```

## 2.5 Inspect Dataset Structure

The column names and data types of each dataset are examined to understand
the attributes available, identify potential key fields, and determine whether
the inferred data types are appropriate for subsequent processing.

```python
# Display column names and data types for each dataset

for name, df in datasets.items():
    print(f"\n{'=' * 70}")
    print(f"DATASET: {name}")
    print(f"{'=' * 70}")
    print(df.dtypes)

```

## 2.6 Preview Source Data

A sample of records from each dataset is reviewed to understand the actual
content, formatting, identifiers, and potential relationships between datasets.

```python
# Display the first 3 records from each dataset

for name, df in datasets.items():
    print(f"\n{'=' * 70}")
    print(f"DATASET: {name}")
    print(f"{'=' * 70}")
    display(df.head(3))
```

## 2.7 Identify Candidate Key Columns

Potential identifier and key columns are identified to understand how the
source datasets may relate to one another. These candidate relationships
will be validated before being used for data integration.

```python
# Identify columns containing "_id"

for name, df in datasets.items():
    id_columns = [col for col in df.columns if "_id" in col]

    print(f"{name:<40} {id_columns}")
```

## 2.8 Key Uniqueness Analysis

Candidate identifier columns are assessed for uniqueness and missing values.
This helps determine whether columns may function as primary keys, foreign keys,
or components of composite keys.

```python
# Check uniqueness and missing values for candidate ID columns

for name, df in datasets.items():
    id_columns = [col for col in df.columns if "_id" in col]

    if id_columns:
        print(f"\n{'=' * 70}")
        print(f"DATASET: {name}")
        print(f"{'=' * 70}")

        for col in id_columns:
            total_rows = len(df)
            unique_values = df[col].nunique()
            missing_values = df[col].isna().sum()

            print(
                f"{col:<25} "
                f"Rows: {total_rows:>8,} | "
                f"Unique: {unique_values:>8,} | "
                f"Missing: {missing_values:>6,}"
            )
```

## 2.9 Create Key Profiling Summary

A structured profiling summary is created for candidate identifier columns,
showing row counts, unique values, missing values, and uniqueness percentage.
This supports the identification of potential primary and foreign keys.

```python
# Create a structured summary of candidate ID columns

key_profile = []

for name, df in datasets.items():
    id_columns = [col for col in df.columns if "_id" in col]

    for col in id_columns:
        total_rows = len(df)
        unique_values = df[col].nunique()
        missing_values = df[col].isna().sum()

        key_profile.append({
            "dataset": name,
            "column": col,
            "rows": total_rows,
            "unique_values": unique_values,
            "missing_values": missing_values,
            "uniqueness_pct": round((unique_values / total_rows) * 100, 2)
        })

key_profile_df = pd.DataFrame(key_profile)

display(key_profile_df)
```

## 2.10 Validate Dataset Relationships

Candidate relationships between datasets are validated by checking whether
foreign key values in child datasets exist in the corresponding parent datasets.

This helps identify potential orphan records and confirms whether the proposed
relationships are suitable for downstream data integration.

```python
# Validate foreign key relationships between datasets

relationships = [
    ("orders_dataset", "customer_id", "customers_dataset", "customer_id"),
    ("order_items_dataset", "order_id", "orders_dataset", "order_id"),
    ("order_items_dataset", "product_id", "products_dataset", "product_id"),
    ("order_items_dataset", "seller_id", "sellers_dataset", "seller_id"),
    ("order_payments_dataset", "order_id", "orders_dataset", "order_id"),
    ("order_reviews_dataset", "order_id", "orders_dataset", "order_id"),
    ("closed_deals_dataset", "mql_id", "marketing_qualified_leads_dataset", "mql_id")
]

for child_table, child_key, parent_table, parent_key in relationships:

    child_values = set(datasets[child_table][child_key].dropna())
    parent_values = set(datasets[parent_table][parent_key].dropna())

    orphan_values = child_values - parent_values

    print(
        f"{child_table}.{child_key} -> "
        f"{parent_table}.{parent_key} | "
        f"Orphan keys: {len(orphan_values):,}"
    )
```

## 2.11 Missing Value Analysis

Missing values are assessed across all source datasets to identify incomplete
attributes and determine the scale of potential data quality issues.

Both the number and percentage of missing values are calculated to distinguish
isolated gaps from systematic data completeness issues.

```python
# Profile missing values across all datasets

missing_profile = []

for name, df in datasets.items():
    for col in df.columns:

        missing_count = df[col].isna().sum()

        if missing_count > 0:
            missing_pct = (missing_count / len(df)) * 100

            missing_profile.append({
                "dataset": name,
                "column": col,
                "missing_count": missing_count,
                "missing_pct": round(missing_pct, 2)
            })

missing_profile_df = pd.DataFrame(missing_profile)

display(
    missing_profile_df.sort_values(
        by="missing_pct",
        ascending=False
    )
)
```

## 2.12 Duplicate Record Analysis

Exact duplicate records are assessed across all source datasets to identify
potential data duplication issues.

The analysis distinguishes complete row duplication from legitimate repetition
of identifiers caused by one-to-many business relationships.

```python
# Check for exact duplicate rows in each dataset

duplicate_profile = []

for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()
    duplicate_pct = (duplicate_count / len(df)) * 100

    duplicate_profile.append({
        "dataset": name,
        "rows": len(df),
        "duplicate_rows": duplicate_count,
        "duplicate_pct": round(duplicate_pct, 2)
    })

duplicate_profile_df = pd.DataFrame(duplicate_profile)

display(
    duplicate_profile_df.sort_values(
        by="duplicate_rows",
        ascending=False
    )
)
```

## 2.13 Investigate Geolocation Duplicates

The geolocation dataset contains a significant number of exact duplicate records.
The duplicated records are examined to understand their characteristics before
determining whether deduplication is appropriate.

```python
# Select the geolocation dataset
geo_df = datasets["geolocation_dataset"]

# Return all records that occur more than once
geo_duplicates = geo_df[
    geo_df.duplicated(keep=False)
]

print(f"Total rows: {len(geo_df):,}")
print(f"Duplicate rows: {geo_df.duplicated().sum():,}")
print(f"Rows involved in duplicate groups: {len(geo_duplicates):,}")

display(geo_duplicates.head(20))
```

## 2.14 Investigate Missing Order Dates

Missing order lifecycle dates are analysed against order status to determine
whether the missing values represent data quality issues or expected business
process behaviour.

```python
# Select the orders dataset
orders_df = datasets["orders_dataset"]

# Analyse missing customer delivery dates by order status
missing_delivery_by_status = (
    orders_df[
        orders_df["order_delivered_customer_date"].isna()
    ]
    ["order_status"]
    .value_counts()
    .reset_index()
)

missing_delivery_by_status.columns = [
    "order_status",
    "missing_delivery_count"
]

display(missing_delivery_by_status)
```

## 2.15 Validate Delivered Orders with Missing Delivery Dates

Orders marked as delivered but without a recorded customer delivery date are
examined as potential data quality exceptions.

```python
# Identify delivered orders with no customer delivery date

delivered_missing_date = orders_df[
    (orders_df["order_status"] == "delivered") &
    (orders_df["order_delivered_customer_date"].isna())
]

print(f"Delivered orders with missing delivery date: {len(delivered_missing_date)}")

display(delivered_missing_date)
```

Numeric attributes are profiled using descriptive statistics to assess their

## 2.17 Business Rule Validation

Selected numeric attributes are tested against expected business rules to
identify invalid or potentially anomalous values.

The checks focus on negative monetary values, zero-value payments, product
measurements, and review score validity.

```python
# Targeted business-rule validation

items_df = datasets["order_items_dataset"]
payments_df = datasets["order_payments_dataset"]
reviews_df = datasets["order_reviews_dataset"]
products_df = datasets["products_dataset"]

validation_results = {
    "Negative product prices":
        (items_df["price"] < 0).sum(),

    "Negative freight values":
        (items_df["freight_value"] < 0).sum(),

    "Negative payment values":
        (payments_df["payment_value"] < 0).sum(),

    "Zero payment values":
        (payments_df["payment_value"] == 0).sum(),

    "Invalid review scores":
        (~reviews_df["review_score"].between(1, 5)).sum(),

    "Zero or negative product weights":
        (products_df["product_weight_g"] <= 0).sum(),

    "Zero or negative product lengths":
        (products_df["product_length_cm"] <= 0).sum(),

    "Zero or negative product heights":
        (products_df["product_height_cm"] <= 0).sum(),

    "Zero or negative product widths":
        (products_df["product_width_cm"] <= 0).sum()
}

validation_df = pd.DataFrame(
    validation_results.items(),
    columns=["validation_rule", "records_found"]
)

display(validation_df)
```

## 2.18 Date and Time Profiling

Date and timestamp attributes are profiled to determine their temporal coverage
and identify potential invalid or inconsistent date values.

Temporary conversions are used for profiling purposes without modifying the
original source datasets.

```python
# Profile date and timestamp columns without modifying the source datasets

date_profile = []

for name, df in datasets.items():

    date_columns = [
        col for col in df.columns
        if "date" in col.lower() or "timestamp" in col.lower()
    ]

    for col in date_columns:

        converted_dates = pd.to_datetime(
            df[col],
            errors="coerce"
        )

        original_non_null = df[col].notna().sum()
        converted_non_null = converted_dates.notna().sum()

        invalid_dates = original_non_null - converted_non_null

        date_profile.append({
            "dataset": name,
            "column": col,
            "min_date": converted_dates.min(),
            "max_date": converted_dates.max(),
            "invalid_dates": invalid_dates
        })

date_profile_df = pd.DataFrame(date_profile)

display(date_profile_df)
```

## 2.19 Phase 2 Summary

The source data assessment identified 11 CSV datasets representing customer,
order, product, seller, payment, review, geolocation, and marketing/sales
entities.

### Key Findings

- Candidate primary and foreign keys were identified and validated across the
  principal datasets.
- No orphan foreign-key values were identified across the tested relationships.
- Missing values were identified across several datasets, with particularly high
  levels in selected closed-deal attributes and optional review text fields.
- Most missing customer delivery dates were consistent with non-delivered order

- The geolocation dataset contained 261,831 exact duplicate rows (26.18%).
  Repeated ZIP-code prefixes with differing coordinates were also observed and
  should not automatically be treated as duplicates.
- No negative product prices, freight values, or payment values were identified.
- 9 zero-value payment records and 4 zero or negative product-weight records
  require further investigation.
- Review scores were within the expected range of 1 to 5.
- All identified date and timestamp values successfully converted to datetime
  during profiling, with no malformed non-null date values detected.
- The principal order data spans September 2016 to October 2018.
- A shipping-limit date extending into 2020 was observed and flagged for
  subsequent validation.

These findings will inform the cleaning, transformation, and validation rules
applied in the next phase.
